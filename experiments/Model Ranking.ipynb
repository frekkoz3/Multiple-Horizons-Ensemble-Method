# Model Ranking

In this notebook, we process the results of model evaluation to create a ranking of model performances that better represents them compared to simply using MSE.

For each dataset, the ranking is computed in two different ways:
* assigning point only to the best performing model for each time series (1st -> 1 point, 2nd, 3rd, 4th -> 0 pts)
* assigning decreasing points to all models, based on performance (1st -> 1 point, 2nd -> 2 pts, 3rd -> 3 pts, 4th -> 4 pts)

The results are then aggregated in some way along the model columns to obtain the final rankings.

```python
#Imports and configs
import pandas as pd

# Paths to the csv containing results, based on original dataset
# WARNING: this formulation depends on the time the results were reated and WILL BREAK if new results are created
PATH_E='../results/errors_e_20260203_151057.csv'
PATH_S='../results/errors_s_20260203_012730.csv'
PATH_T='../results/errors_t_20260202_234253.csv'
PATH_V='../results/errors_v_20260203_002115.csv'
PATH_W='../results/errors_w_20260202_232714.csv'
```

## Electricity Dataset Results

```python
# Expected columns: Method, ID, Error
df = pd.read_csv(PATH_E)

to_exclude = ["mean", "whole_dataset"]
df_filtered = df[~df['ID'].isin(to_exclude)].copy()

# Pivot the data: move 'Method' to columns and 'ID' to rows
df_pivot = df_filtered.pivot(index='ID', columns='Method', values='Error')

#  Drop rows where no model has data
df_pivot = df_pivot.dropna(how='all', axis=0)

# Ensure data is numeric
df_pivot = df_pivot.apply(pd.to_numeric, errors='coerce')

# --- RANKING SYSTEM 1: WIN COUNTS ---
winners = df_pivot.idxmin(axis=1)
win_counts = winners.value_counts().reindex(df_pivot.columns, fill_value=0)
win_ranking = win_counts.sort_values(ascending=False).reset_index()
win_ranking.columns = ['Model', 'Wins']

# --- RANKING SYSTEM 2: AVERAGE POSITIONS ---
# 1.0 is 1st place, 4.0 is 4th place
ranks_per_series = df_pivot.rank(axis=1, ascending=True, method='average')
avg_rank = ranks_per_series.mean().sort_values().reset_index()
avg_rank.columns = ['Model', 'Average Rank']


print("\n--- Ranking by Wins (Higher is better) ---")
print(win_ranking)

print("\n--- Ranking by Average Position (Lower is better) ---")
print(avg_rank)
```
